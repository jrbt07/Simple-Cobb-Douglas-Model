```python
import sympy as sp

#Welcoming messages
print("Hello! Welcome the the Cobb-Douglas optimizer!")
print("The purpose of this program is to tell you the amount of two goods (selected by you) that you should buy depending on how much you value each good and how much money you want to spend on them.")
print("The program will use a Cobb-Douglas utility function to help us determine the optimal amount of each good you specify.")
print("You don't need to know how this model works to use it, so don't worry if you don't know anything about it.")
print("First, let me know the goods you want to buy. Make sure you don't repeat the good you chose.\n")

#Checking whether the user used the same good twice
def checkGoods(good1, good2):
    if good1 == good2:
        return False
    else: 
        return True

#Checking if the user entered a word and not a number
def checkNumericalInput(userInput):
    try:
        float(userInput) / 2
        return True
    except ValueError:
        return False

#Checking if the user entered a valid rating [1,10]
def checkPreferenceRange(rating):
    if rating > 10 or rating < 1:
        return False
    else:
        return True

#Checking if the user entered a price greater than 0
def checkPriceInput(priceOfGood):
    if priceOfGood <= 0:
        return False
    else:
        return True

#Checking if the budget is greater than 0
def checkBudgetInput(priceGood1, priceGood2, budget):
    if budget <= 0:
        return False
    else:
        return True

#Normalizing the first rating to get the first exponent
def getExponent1(ratingGood1, ratingGood2):
    ratingSum = ratingGood1 + ratingGood2
    exp1 = ratingGood1 / ratingSum
    return exp1

#Normalizing the second rating to get the second exponent
def getExponent2(ratingGood1, ratingGood2):
    ratingSum = ratingGood1 + ratingGood2
    exp2 = ratingGood2 / ratingSum
    return exp2

#Using the exponents to set up the Cobb Douglas function
def getCobbDouglasFunction(expGood1, expGood2):
    #Setting up x and y as 'mathematical' variables
    x = sp.symbols("x")
    y = sp.symbols("y")
    #x is the amount of good1, y is the amount of good 2  
    f = (x**expGood1) * (y**expGood2)
    return f

#Using the allocated budget and each good's price to set up the constraint function
def getConstraintFunction(priceGood1, priceGood2, budget):
    g = priceGood1*x + priceGood2*y - budget
    return g
```
